```python
df_oders=spark.createDataFrame([(2,'laptop','2020-12-01'),(37,'Mobile','2020-12-02'),(77,'TV','2020-12-03')],['CustomerID','Ordername','OrderDate'])
display(df_oders)
```

```python
df_customers=spark.createDataFrame([(1,'krishna','maria','Germany'),(2,'rama','ana','mexico'),(3,'sai','antonio','mexico')],['CustomerID','CustomerName','ContactName','Country'])
display(df_customers)
```

```python
df_full=df_oders.join(df_customers,df_oders["CustomerID"]==df_customers["CustomerID"],"full_outer")
df_full.show()
```

```python
from pyspark.sql.functions import col
df_inner=df_oders.alias("a").join(df_customers.alias("b"),col("a.CustomerID")==col("b.CustomerID"),"inner")
df_inner.show()
```

```python
df_test=df_oders.alias("a").join(df_customers.alias("b"),col("a.CustomerID")==col("b.CustomerID"),"left_outer").select(col("a.*"),col("b.CustomerName"),col("b.ContactName"),col("b.country"))
display(df_test)
```

```python
df_inner=df_oders.alias("a").join(df_customers.alias("b"),col("a.CustomerID")==col("b.CustomerID"),"left_outer")
df_inner.show()
```

```python
df_inner=df_oders.alias("a").join(df_customers.alias("b"),col("a.CustomerID")==col("b.CustomerID"),"left_semi")
df_inner.show()
```

```python
df_inner=df_oders.alias("a").join(df_customers.alias("b"),col("a.CustomerID")==col("b.CustomerID"),"inner")
df_inner.show()
```
