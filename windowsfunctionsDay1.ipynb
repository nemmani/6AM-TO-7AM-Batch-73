```python
%sql
--use case 1 - row number
--give a unique number to employees within each department based on salary

select emp_id, emp_name, dept, salary, row_number() over(partition by dept order by salary desc)as row_num from employees
```

```python
%sql
select emp_id, emp_name, dept, salary, rank() over(partition by dept order by salary desc)as salary_rank from employees
```

```python
%sql
select emp_id, emp_name, dept, salary, dense_rank() over(partition by dept order by salary desc)as salary_denserank from employees
```

```python
%sql
--use case
--  find top 2 employees from each department
-- CTE is a temporary named result set that you create using with.
-- It exists only for the duration of the sql query

with ranked_employees as (
    select emp_id, emp_name, dept, salary, row_number() over(partition by dept order by salary desc)as rn from employees
)
select * from ranked_employees where rn <=2
```
